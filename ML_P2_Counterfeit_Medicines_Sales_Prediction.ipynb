# Decision Tree

```python
# Import RandomizedSearchCV and DecisionTreeRegressor for hyperparameter tuning
from sklearn.model_selection import RandomizedSearchCV
from sklearn.tree import DecisionTreeRegressor
```

```python
# Define a helper function to print the top-N hyperparameter search results
def report(results, n_top=3):
    for i in range(1, n_top + 1):
        candidates = np.flatnonzero(results['rank_test_score'] == i)
        for candidate in candidates:
            print("Model with rank: {0}".format(i))
            print("Mean validation score: {0:.5f} (std: {1:.5f})".format(
                  results['mean_test_score'][candidate],
                  results['std_test_score'][candidate]))
            print("Parameters: {0}".format(results['params'][candidate]))
            print("")
```

```python
# Define the hyperparameter grid to search over for the Decision Tree
paramsdt={ 
        'max_depth':[None,5,10,15,20,30,50,70],
            'min_samples_leaf':[1,2,5,10,15,20], 
            'min_samples_split':[2,5,10,15,20]
       }
```

```python
# Instantiate a base Decision Tree Regressor
regdt=DecisionTreeRegressor()
```

```python
# Set up randomized search cross-validation (10-fold) to tune the Decision Tree using MAE as the scoring metric
random_search=RandomizedSearchCV(regdt,
                                 cv=10,
                                 param_distributions=paramsdt,
                                 scoring='neg_mean_absolute_error',
                                 n_iter=10,n_jobs=-1,verbose=20
                                    )
```

```python
# Run the randomized search, fitting the Decision Tree across hyperparameter combinations
random_search.fit(x_train,y_train)
```

```python
# Print the top 5 hyperparameter combinations and their validation scores
report(random_search.cv_results_,5)
```

```python
# Display the best-performing Decision Tree estimator found by the search
random_search.best_estimator_
```

```python
# Display the best cross-validated score (negative MAE) from the search
random_search.best_score_
```

```python
# Convert the best score into a positive mean absolute error (MAE) value
mae=abs(random_search.best_score_)
```

```python
# Retrain the best Decision Tree estimator on the full training data
et=random_search.best_estimator_
et.fit(x_train,y_train)
```

```python
# Calculate a normalized accuracy-like score relative to the sales value range
Scoredt=1-(mae/1660)
print(Scoredt)
```

```python
# Compute and sort feature importances from the trained Decision Tree
feat_imp_df=pd.DataFrame({'features':x_train.columns,'importance':et.feature_importances_})

feat_imp_df.sort_values('importance',ascending=False)
```

```python
# Display test dataset info before generating final predictions
bd_test.info()
```

```python
# Generate Counterfeit_Sales predictions on the test set using the tuned Decision Tree
submissions_dt=pd.DataFrame({target:et.predict(bd_test)})
```

```python
# Combine the test Medicine_IDs with the Decision Tree predictions for submission
Final_submit_dt=pd.concat([med_ID_test,submissions_dt],axis=1)
```

```python
# Save the Decision Tree predictions to a submission CSV file
Final_submit_dt.to_csv('sample_submission_dt.csv',index=False)
```
